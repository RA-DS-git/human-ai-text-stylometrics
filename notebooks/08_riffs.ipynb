# 08 — Riffs (Combined Visualizations)
**Inputs:** `data/docs_sampled.csv`, `data/hc3_DOC_SENT.csv`, `data/hc3_LDA_THETA.csv`, `data/hc3_LDA_PHI.csv`

**Images:** `images/riff1_sentiment_heatmap.png`, `images/riff2_length_boxplot.png`, `images/riff3_topic_heatmap.png`

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import warnings
warnings.filterwarnings('ignore')

sns.set_theme(style='whitegrid')
COLORS = {'human':'#4C72B0', 'chatgpt':'#DD8452'}
print('Libraries loaded.')
```

```python
docs_sampled = pd.read_csv('../data/docs_sampled.csv')
DOC_SENT= pd.read_csv('../data/hc3_DOC_SENT.csv', index_col='doc_id')
LDA_THETA= pd.read_csv('../data/hc3_LDA_THETA.csv', index_col='doc_id')
LDA_PHI = pd.read_csv('../data/hc3_LDA_PHI.csv', index_col='topic')
meta = docs_sampled.set_index('doc_id')[['author_type','source']]
topic_cols = [c for c in LDA_THETA.columns if c.startswith('T')]

print('All data loaded.')
```

## Riff 1 — Sentiment Heatmap by Source × Author Type

```python
heatmap_data = (
    DOC_SENT.groupby(['source','author_type'])['mean_sentiment']
    .mean().unstack()
)

fig, ax = plt.subplots(figsize=(8, 5))
sns.heatmap(
    heatmap_data, annot=True, fmt='.3f',
    cmap='RdYlGn',center=0,
    linewidths=0.5, ax=ax,
    cbar_kws={'label': 'Mean Sentiment Score'}
)
ax.set_title('Riff 1: Mean Sentiment Score by Source Domain and Author Type', fontsize=12)
ax.set_xlabel('Author Type')
ax.set_ylabel('Source Domain')
plt.tight_layout()
plt.savefig('../images/riff1_sentiment_heatmap.png', dpi=150, bbox_inches='tight')
plt.show()
print('Saved images/riff1_sentiment_heatmap.png')
```

## Riff 2 — Document Length Box Plot by Source × Author Type

```python
fig, ax =plt.subplots(figsize=(12, 6))
sns.boxplot(
    data=docs_sampled[docs_sampled['word_count']<= 600],
    x='source', y='word_count', hue='author_type',
    palette=COLORS, ax=ax, width=0.6, fliersize=3
)
ax.set_title('Riff 2: Document Length Distribution by Source Domain and Author Type', fontsize=12)
ax.set_xlabel('Source Domain')
ax.set_ylabel('Word Count (capped at 600)')
ax.tick_params(axis='x', rotation=15)
ax.legend(title='Author type')
plt.tight_layout()
plt.savefig('../images/riff2_length_boxplot.png', dpi=150, bbox_inches='tight')
plt.show()
print('Saved images/riff2_length_boxplot.png')
```

## Riff 3 — Topic Weight Heatmap by Author Type

```python
theta_meta= LDA_THETA.join(meta)
topic_by_author =theta_meta.groupby('author_type')[topic_cols].mean()

topic_by_author.columns = [
    f"{c}\n({LDA_PHI.loc[c].nlargest(1).index[0]})" for c in topic_cols
]

fig, ax = plt.subplots(figsize=(16, 3))
sns.heatmap(
    topic_by_author, annot=True, fmt='.3f',
    cmap='Blues', linewidths=0.5, ax=ax,
    cbar_kws={'label': 'Mean Topic Weight'}
)
ax.set_title('Riff 3: Mean LDA Topic Weight by Author Type', fontsize=12)
ax.set_ylabel('Author Type')
ax.tick_params(axis='x', rotation=15)
plt.tight_layout()
plt.savefig('../images/riff3_topic_heatmap.png', dpi=150, bbox_inches='tight')
plt.show()
print('Saved images/riff3_topic_heatmap.png')
```
